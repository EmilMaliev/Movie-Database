# Data preparation

```python
def prepare_explicit_interactions(
    reviews: pd.DataFrame,
    min_user_ratings: int = 2,
    min_movie_ratings: int = 2,
    keep_latest: bool = True
) -> pd.DataFrame:
    """

    Параметры
    ---------
    reviews : pd.DataFrame
        Исходный датафрейм отзывов.
    min_user_ratings : int
        Минимум оценённых фильмов у пользователя.
    min_movie_ratings : int
        Минимум пользователей, оценивших фильм.
    keep_latest : bool
        Если True, при дублях (user, movie) оставляем последнюю запись по updated_at.

    Возвращает
    ----------
        Таблица с колонками:
        - user_id
        - movie_id
        - rating
        - updated_at
    """

    df = reviews.copy()
    df = df[df["author_rating"].notna()].copy()

    cols = ["author_username", "movie_id", "author_rating", "updated_at"]
    df = df[cols].copy()

    df["author_username"] = df["author_username"].astype(str)
    df["movie_id"] = df["movie_id"].astype(int)
    df["author_rating"] = df["author_rating"].astype(float)
    df["updated_at"] = pd.to_datetime(df["updated_at"], errors="coerce")

    df = df[df["author_username"].str.strip() != ""].copy()
    df = df[df["movie_id"].notna()].copy()

    # Обрабатываем дубликаты 
    if keep_latest:
        df = df.sort_values("updated_at")
        df = df.drop_duplicates(subset=["author_username", "movie_id"], keep="last")
    else:
        # Альтернатива: усреднение рейтингов по дублям
        df = (
            df.groupby(["author_username", "movie_id"], as_index=False)
              .agg({
                  "author_rating": "mean",
                  "updated_at": "max"
              })
        )

    user_counts = df.groupby("author_username")["movie_id"].nunique()
    valid_users = user_counts[user_counts >= min_user_ratings].index
    df = df[df["author_username"].isin(valid_users)].copy()

    movie_counts = df.groupby("movie_id")["author_username"].nunique()
    valid_movies = movie_counts[movie_counts >= min_movie_ratings].index
    df = df[df["movie_id"].isin(valid_movies)].copy()

    user_counts = df.groupby("author_username")["movie_id"].nunique()
    valid_users = user_counts[user_counts >= min_user_ratings].index
    df = df[df["author_username"].isin(valid_users)].copy()

    df = df.rename(columns={
        "author_username": "user_id",
        "author_rating": "rating"
    })

    df = df.sort_values(["user_id", "updated_at", "movie_id"]).reset_index(drop=True)

    return df

interactions = prepare_explicit_interactions(
    reviews,
    min_user_ratings=2,
    min_movie_ratings=2,
    keep_latest=True
)

print(interactions.shape)
print(interactions.head())
print("Users:", interactions["user_id"].nunique())
print("Movies:", interactions["movie_id"].nunique())
print("Ratings:", len(interactions))
```

# Split by users

```python
def split_by_user(
    interactions: pd.DataFrame,
    user_col: str = "user_id",
    item_col: str = "movie_id",
    test_frac: float = 0.2,
    seed: int = 42
) -> tuple[pd.DataFrame, pd.DataFrame]:
    """
    Делит на train/test по пользователям

    Логика:
    - для каждого пользователя перемешиваем его взаимодействия
    - test_size = max(1, round(n * test_frac))
    - при n >= 2 гарантируем:
        * минимум 1 объект в test
        * минимум 1 объект в train

    Параметры
    ---------
    interactions : pd.DataFrame
        Таблица взаимодействий, например:
        user_id, movie_id, rating, updated_at
    user_col : str
        Название колонки пользователя
    item_col : str
        Название колонки объекта
    test_frac : float
        Доля объектов пользователя, которые пойдут в test
    seed : int
        Random seed для воспроизводимости

    """

    train_parts = []
    test_parts = []

    for user_id, group in interactions.groupby(user_col):
        group = group.sample(frac=1, random_state=seed).reset_index(drop=True)
        n = len(group)

        # Пользователь с 1 взаимодействием оставляем в train
        if n < 2:
            train_parts.append(group)
            continue

        test_size = max(1, int(round(n * test_frac)))

        # Чтобы точно остался хотя бы 1 объект в train
        test_size = min(test_size, n - 1)

        test_part = group.iloc[:test_size].copy()
        train_part = group.iloc[test_size:].copy()

        train_parts.append(train_part)
        test_parts.append(test_part)

    train_df = pd.concat(train_parts, ignore_index=True)
    test_df = pd.concat(test_parts, ignore_index=True) if len(test_parts) > 0 else pd.DataFrame(columns=interactions.columns)

    return train_df, test_df

train_df, test_df = split_by_user(
    interactions,
    user_col="user_id",
    item_col="movie_id",
    test_frac=0.2,
    seed=42
)

print("Train shape:", train_df.shape)
print("Test shape:", test_df.shape)
print("Train users:", train_df["user_id"].nunique())
print("Test users:", test_df["user_id"].nunique())
```

# user-item matrix

```python
def build_cf_data(
    train_df: pd.DataFrame,
    user_col: str = "user_id",
    item_col: str = "movie_id",
    rating_col: str = "rating"
) -> dict:

    df = train_df.copy()

    users = sorted(df[user_col].unique())
    items = sorted(df[item_col].unique())

    user2idx = {u: i for i, u in enumerate(users)}
    idx2user = {i: u for u, i in user2idx.items()}

    item2idx = {it: j for j, it in enumerate(items)}
    idx2item = {j: it for it, j in item2idx.items()}

    row_idx = df[user_col].map(user2idx).to_numpy()
    col_idx = df[item_col].map(item2idx).to_numpy()
    values = df[rating_col].astype(np.float32).to_numpy()

    R = csr_matrix(
        (values, (row_idx, col_idx)),
        shape=(len(users), len(items)),
        dtype=np.float32
    )

    train_seen = (
        df.groupby(user_col)[item_col]
          .apply(set)
          .to_dict()
    )

    user_means_series = df.groupby(user_col)[rating_col].mean()
    user_means = user_means_series.to_dict()

    return {
        "train_seen": train_seen,
        "user2idx": user2idx,
        "idx2user": idx2user,
        "item2idx": item2idx,
        "idx2item": idx2item,
        "R": R,
        "user_means": user_means,
    }

cf_data = build_cf_data(
    train_df=train_df,
    user_col="user_id",
    item_col="movie_id",
    rating_col="rating"
)

R = cf_data["R"]
train_seen = cf_data["train_seen"]
user2idx = cf_data["user2idx"]
item2idx = cf_data["item2idx"]

print("R shape:", R.shape)
print("R numof not zeros:", R.nnz)  
print("Users:", len(user2idx))
print("Movies:", len(item2idx))

density = R.nnz / (R.shape[0] * R.shape[1])
print("Matrix density:", density)

example_user = next(iter(train_seen))
print("Example user:", example_user)
print("Seen movies count:", len(train_seen[example_user]))
print("User mean rating:", cf_data["user_means"][example_user])
```

# Item CF

```python
def keep_top_k_sparse_rows(matrix: csr_matrix, k: int, positive_only: bool = True) -> csr_matrix:

    matrix = matrix.tocsr()

    new_data = []
    new_indices = []
    new_indptr = [0]

    for row_idx in range(matrix.shape[0]):
        start, end = matrix.indptr[row_idx], matrix.indptr[row_idx + 1]
        row_data = matrix.data[start:end]
        row_cols = matrix.indices[start:end]

        if positive_only:
            mask = row_data > 0
            row_data = row_data[mask]
            row_cols = row_cols[mask]

        if len(row_data) == 0:
            new_indptr.append(len(new_data))
            continue

        if len(row_data) > k:
            top_pos = np.argpartition(row_data, -k)[-k:]
            top_order = np.argsort(row_data[top_pos])[::-1]
            selected = top_pos[top_order]
            row_data = row_data[selected]
            row_cols = row_cols[selected]
        else:
            order = np.argsort(row_data)[::-1]
            row_data = row_data[order]
            row_cols = row_cols[order]

        new_data.extend(row_data.tolist())
        new_indices.extend(row_cols.tolist())
        new_indptr.append(len(new_data))

    return csr_matrix(
        (
            np.array(new_data, dtype=np.float32),
            np.array(new_indices, dtype=np.int32),
            np.array(new_indptr, dtype=np.int32),
        ),
        shape=matrix.shape,
    )
```

```python
def build_itemcf_model(
    train_df: pd.DataFrame,
    cf_data: dict,
    user_col: str = "user_id",
    item_col: str = "movie_id",
    rating_col: str = "rating",
    top_k_neighbors: int = 50
) -> dict:
    """
    Использует:
    - сырую матрицу R из cf_data
    - центрированную по пользователю матрицу R_centered
    - cosine similarity между фильмами
    """

    df = train_df.copy()

    user2idx = cf_data["user2idx"]
    idx2user = cf_data["idx2user"]
    item2idx = cf_data["item2idx"]
    idx2item = cf_data["idx2item"]
    train_seen = cf_data["train_seen"]
    R = cf_data["R"]
    user_means = cf_data["user_means"]

    # Центрируем рейтинги: rating - mean_user_rating
    df["user_mean"] = df[user_col].map(user_means)
    df["rating_centered"] = (df[rating_col] - df["user_mean"]).astype(np.float32)

    row_idx = df[user_col].map(user2idx).to_numpy()
    col_idx = df[item_col].map(item2idx).to_numpy()
    values = df["rating_centered"].to_numpy(dtype=np.float32)

    R_centered = csr_matrix(
        (values, (row_idx, col_idx)),
        shape=R.shape,
        dtype=np.float32
    )

    item_sim = cosine_similarity(R_centered.T, dense_output=False)
    item_sim = item_sim.tocsr()

    item_sim.setdiag(0.0)
    item_sim.eliminate_zeros()

    if top_k_neighbors is not None:
        item_sim = keep_top_k_sparse_rows(item_sim, k=top_k_neighbors, positive_only=True)

    return {
        "R": R,
        "R_centered": R_centered,
        "item_sim": item_sim,
        "train_seen": train_seen,
        "user_means": user_means,
        "user2idx": user2idx,
        "idx2user": idx2user,
        "item2idx": item2idx,
        "idx2item": idx2item,
        "top_k_neighbors": top_k_neighbors,
    }
```

```python
def recommend_itemcf_for_user(
    user_id,
    model: dict,
    top_n: int = 100
) -> pd.DataFrame:
    """
    Генерирует кандидатов для пользователя через item-item CF.

    Возвращает DataFrame с колонками:
    - movie_id
    - itemcf_score
    - sim_sum
    - n_contrib
    """

    user2idx = model["user2idx"]
    idx2item = model["idx2item"]
    item_sim = model["item_sim"]
    R = model["R"]
    train_seen = model["train_seen"]
    user_means = model["user_means"]

    if user_id not in user2idx:
        return pd.DataFrame(columns=["movie_id", "itemcf_score", "sim_sum", "n_contrib"])

    u_idx = user2idx[user_id]
    user_row = R.getrow(u_idx)

    if user_row.nnz == 0:
        return pd.DataFrame(columns=["movie_id", "itemcf_score", "sim_sum", "n_contrib"])

    rated_item_indices = user_row.indices
    rated_values = user_row.data
    user_mean = user_means.get(user_id, float(np.mean(rated_values)))

    # фильмы выше среднего рейтинга пользователя
    seed_mask = rated_values > user_mean

    # Если таких нет, используем всю историю
    if seed_mask.sum() == 0:
        seed_mask = np.ones_like(rated_values, dtype=bool)

    seed_item_indices = rated_item_indices[seed_mask]
    seed_weights = rated_values[seed_mask]

    seen_movies = train_seen.get(user_id, set())

    score_num = defaultdict(float)
    score_den = defaultdict(float)
    n_contrib = defaultdict(int)

    for seed_idx, seed_weight in zip(seed_item_indices, seed_weights):
        sim_row = item_sim.getrow(seed_idx)

        for cand_idx, sim_val in zip(sim_row.indices, sim_row.data):
            cand_movie_id = idx2item[cand_idx]

            if cand_movie_id in seen_movies:
                continue
            if sim_val <= 0:
                continue

            score_num[cand_idx] += float(sim_val * seed_weight)
            score_den[cand_idx] += float(sim_val)
            n_contrib[cand_idx] += 1

    if len(score_num) == 0:
        return pd.DataFrame(columns=["movie_id", "itemcf_score", "sim_sum", "n_contrib"])

    rows = []
    for cand_idx in score_num:
        final_score = score_num[cand_idx] / score_den[cand_idx] if score_den[cand_idx] > 0 else 0.0
        rows.append({
            "movie_id": idx2item[cand_idx],
            "itemcf_score": final_score,
            "sim_sum": score_den[cand_idx],
            "n_contrib": n_contrib[cand_idx],
        })

    recs = pd.DataFrame(rows)
    recs = recs.sort_values(
        ["itemcf_score", "sim_sum", "n_contrib", "movie_id"],
        ascending=[False, False, False, True]
    ).head(top_n).reset_index(drop=True)

    return recs
```

```python
def build_popular_movies_model(
    train_df: pd.DataFrame,
    item_col: str = "movie_id",
    rating_col: str = "rating",
    min_ratings: int = 5
) -> pd.DataFrame:
    """
    Оставляет фильмы, у которых rating_count >= min_ratings,
    """

    popular_model = (
        train_df.groupby(item_col)[rating_col]
        .agg(mean_rating="mean", rating_count="count")
        .reset_index()
    )

    popular_model = popular_model[popular_model["rating_count"] >= min_ratings].copy()

    popular_model = popular_model.sort_values(
        ["mean_rating", "rating_count", item_col],
        ascending=[False, False, True]
    ).reset_index(drop=True)

    return popular_model
```

```python
def recommend_popular_for_user(
    user_id,
    popular_model: pd.DataFrame,
    train_seen: dict,
    item_col: str = "movie_id",
    top_n: int = 100
) -> pd.DataFrame:
    """
    Возвращает top-N популярных фильмов для пользователя,
    исключая уже просмотренные/оценённые.
    """

    seen_movies = train_seen.get(user_id, set())

    recs = popular_model[~popular_model[item_col].isin(seen_movies)].copy()
    recs = recs.head(top_n).reset_index(drop=True)

    recs = recs.rename(columns={"mean_rating": "popular_score"})

    return recs[[item_col, "popular_score", "rating_count"]]
```

```python
def recommend_itemcf_with_fallback(
    user_id,
    itemcf_model: dict,
    popular_model: pd.DataFrame,
    top_n: int = 100
) -> pd.DataFrame:
    """
    Сначала рекомендует через item-CF.
    Если кандидатов меньше top_n, добивает popularity fallback.
    """

    train_seen = itemcf_model["train_seen"]

    rec_cf = recommend_itemcf_for_user(
        user_id=user_id,
        model=itemcf_model,
        top_n=top_n
    ).copy()

    rec_cf["source"] = "itemcf"

    if len(rec_cf) >= top_n:
        return rec_cf.head(top_n).reset_index(drop=True)

    needed = top_n - len(rec_cf)

    rec_pop = recommend_popular_for_user(
        user_id=user_id,
        popular_model=popular_model,
        train_seen=train_seen,
        item_col="movie_id",
        top_n=top_n * 2
    ).copy()

    # Убираем дубли с тем, что уже пришло из item-CF
    cf_movies = set(rec_cf["movie_id"].tolist())
    rec_pop = rec_pop[~rec_pop["movie_id"].isin(cf_movies)].copy()

    rec_pop = rec_pop.head(needed).reset_index(drop=True)
    rec_pop["source"] = "popular_fallback"

    if "itemcf_score" not in rec_pop.columns:
        rec_pop["itemcf_score"] = 0.0
    if "sim_sum" not in rec_pop.columns:
        rec_pop["sim_sum"] = 0.0
    if "n_contrib" not in rec_pop.columns:
        rec_pop["n_contrib"] = 0

    result = pd.concat([rec_cf, rec_pop], ignore_index=True)

    # Сохраняем приоритет item-CF выше fallback
    result["source_priority"] = result["source"].map({
        "itemcf": 0,
        "popular_fallback": 1
    })

    result = result.sort_values(
        ["source_priority", "itemcf_score", "popular_score", "sim_sum", "rating_count", "movie_id"],
        ascending=[True, False, False, False, False, True]
    ).head(top_n).reset_index(drop=True)

    result = result.drop(columns=["source_priority"], errors="ignore")

    return result
```

```python
cf_data = build_cf_data(
    train_df=train_df,
    user_col="user_id",
    item_col="movie_id",
    rating_col="rating"
)

itemcf_model = build_itemcf_model(
    train_df=train_df,
    cf_data=cf_data,
    user_col="user_id",
    item_col="movie_id",
    rating_col="rating",
    top_k_neighbors=50
)

example_user = train_df["user_id"].iloc[0]

rec_itemcf = recommend_itemcf_for_user(
    user_id=example_user,
    model=itemcf_model,
    top_n=5
)

movies['id'] = pd.to_numeric(movies['id'], errors="coerce")

rec_itemcf_named = rec_itemcf.merge(
    movies[["id", "title"]],
    left_on="movie_id",
    right_on="id",
    how="left"
)

print(rec_itemcf_named[["movie_id", "title", "itemcf_score", "sim_sum", "n_contrib"]])
```

# Метрики

1. Recall@K
$$
Recall@K = \frac{\left| Relevant_u \cap Recommended_u^K \right|}{\left| Relevant_u \right|}
$$

2. HitRate@K

Для каждого пользователя:

- 1, если хотя бы один релевантный фильм попал в top-K
- 2, 0, если ни один не попал

```python
def build_test_relevant(
    test_df: pd.DataFrame,
    user_col: str = "user_id",
    item_col: str = "movie_id",
    rating_col: str = "rating",
    relevant_threshold: float = 7.0
) -> dict:
    """
    Строит словарь релевантных объектов из test:
    user_id -> set(movie_id), где rating >= relevant_threshold
    """

    rel_df = test_df[test_df[rating_col] >= relevant_threshold].copy()

    test_relevant = (
        rel_df.groupby(user_col)[item_col]
        .apply(set)
        .to_dict()
    )

    return test_relevant
```

```python
def evaluate_itemcf_candidates(
    model: dict,
    test_relevant: dict,
    top_k: int = 100
) -> tuple[dict, pd.DataFrame]:
    """
    Оценивает Stage 1 candidate generation через Recall@K и HitRate@K.

    Параметры
    ---------
    model : dict
        itemcf_model
    test_relevant : dict
        user_id -> set(relevant movie_id from test)
    top_k : int
        сколько кандидатов генерировать на пользователя

    """

    rows = []

    for user_id, relevant_items in test_relevant.items():
        if len(relevant_items) == 0:
            continue

        recs = recommend_itemcf_for_user(
            user_id=user_id,
            model=model,
            top_n=top_k
        )

        candidate_items = set(recs["movie_id"].tolist())
        hits = relevant_items & candidate_items

        recall = len(hits) / len(relevant_items) if len(relevant_items) > 0 else np.nan
        hitrate = 1.0 if len(hits) > 0 else 0.0

        rows.append({
            "user_id": user_id,
            "n_relevant": len(relevant_items),
            "n_candidates": len(candidate_items),
            "n_hits": len(hits),
            "recall_at_k": recall,
            "hitrate_at_k": hitrate,
        })

    per_user_df = pd.DataFrame(rows)

    if len(per_user_df) == 0:
        summary_metrics = {
            "users_evaluated": 0,
            "mean_recall_at_k": np.nan,
            "mean_hitrate_at_k": np.nan,
            "median_recall_at_k": np.nan,
            "avg_candidates_generated": np.nan,
        }
        return summary_metrics, per_user_df

    summary_metrics = {
        "users_evaluated": int(len(per_user_df)),
        "mean_recall_at_k": float(per_user_df["recall_at_k"].mean()),
        "mean_hitrate_at_k": float(per_user_df["hitrate_at_k"].mean()),
        "median_recall_at_k": float(per_user_df["recall_at_k"].median()),
        "avg_candidates_generated": float(per_user_df["n_candidates"].mean()),
    }

    return summary_metrics, per_user_df
```

```python
test_relevant = build_test_relevant(
    test_df=test_df,
    user_col="user_id",
    item_col="movie_id",
    rating_col="rating",
    relevant_threshold=6.0
)

summary_metrics, per_user_metrics = evaluate_itemcf_candidates(
    model=itemcf_model,
    test_relevant=test_relevant,
    top_k=100
)

print(summary_metrics)
per_user_metrics.head(20)
```

```python
def evaluate_itemcf_with_fallback(
    itemcf_model: dict,
    popular_model: pd.DataFrame,
    test_relevant: dict,
    top_k: int = 100
) -> tuple[dict, pd.DataFrame]:

    rows = []

    for user_id, relevant_items in test_relevant.items():
        if len(relevant_items) == 0:
            continue

        recs = recommend_itemcf_with_fallback(
            user_id=user_id,
            itemcf_model=itemcf_model,
            popular_model=popular_model,
            top_n=top_k
        )

        candidate_items = set(recs["movie_id"].tolist())
        hits = relevant_items & candidate_items

        n_itemcf = int((recs["source"] == "itemcf").sum()) if "source" in recs.columns else np.nan
        n_fallback = int((recs["source"] == "popular_fallback").sum()) if "source" in recs.columns else np.nan

        recall = len(hits) / len(relevant_items) if len(relevant_items) > 0 else np.nan
        hitrate = 1.0 if len(hits) > 0 else 0.0
        precision = len(hits) / len(candidate_items) if len(candidate_items) > 0 else 0.0

        rows.append({
            "user_id": user_id,
            "n_relevant": len(relevant_items),
            "n_candidates": len(candidate_items),
            "n_hits": len(hits),
            "recall_at_k": recall,
            "hitrate_at_k": hitrate,
            "precision_at_k": precision,
            "n_itemcf": n_itemcf,
            "n_fallback": n_fallback,
        })

    per_user_df = pd.DataFrame(rows)

    if len(per_user_df) == 0:
        summary_metrics = {
            "users_evaluated": 0,
            "mean_recall_at_k": np.nan,
            "median_recall_at_k": np.nan,
            "mean_hitrate_at_k": np.nan,
            "mean_precision_at_k": np.nan,
            "avg_candidates_generated": np.nan,
            "avg_itemcf_candidates": np.nan,
            "avg_fallback_candidates": np.nan,
            "share_full_topk": np.nan,
        }
        return summary_metrics, per_user_df

    summary_metrics = {
        "users_evaluated": int(len(per_user_df)),
        "mean_recall_at_k": float(per_user_df["recall_at_k"].mean()),
        "median_recall_at_k": float(per_user_df["recall_at_k"].median()),
        "mean_hitrate_at_k": float(per_user_df["hitrate_at_k"].mean()),
        "mean_precision_at_k": float(per_user_df["precision_at_k"].mean()),
        "avg_candidates_generated": float(per_user_df["n_candidates"].mean()),
        "avg_itemcf_candidates": float(per_user_df["n_itemcf"].mean()),
        "avg_fallback_candidates": float(per_user_df["n_fallback"].mean()),
        "share_full_topk": float((per_user_df["n_candidates"] == top_k).mean()),
    }

    return summary_metrics, per_user_df
```

```python
test_relevant = build_test_relevant(
    test_df=test_df,
    user_col="user_id",
    item_col="movie_id",
    rating_col="rating",
    relevant_threshold=6.0
)

summary_fallback, per_user_fallback = evaluate_itemcf_with_fallback(
    itemcf_model=itemcf_model,
    popular_model=popular_model,
    test_relevant=test_relevant,
    top_k=100
)

print(summary_fallback)
per_user_fallback.head(20)
```
