```python
def get_reccomendations(df:pd.DataFrame,id_list:list,index:str):
  tmp=df[df['id'].isin(id_list)].drop('id',axis=1).to_numpy()
  tmp=np.mean(tmp, axis=0)
  dim=tmp.shape[0]
  u = AnnoyIndex(dim, 'euclidean')
  u.load(index)
  indices = u.get_nns_by_vector(query, 10)
  ids=df['id'].iloc[indices].tolist()
  return  np.setdiff1d(ids, np.intersect1d(ids, id_list))

ids=get_reccomendations(movie_df_full,["2","12","115"],"/content/test.ann")
movies_df[movies_df['id'].isin(ids)]
```

```python
movie_voc={reviews['movie_id'].unique().tolist()[i]:i for i in range(len(reviews['movie_id'].unique().tolist()))}
```

```python
author_username_voc={reviews['author_username'].unique().tolist()[i]:i for i in range(len(reviews['author_username'].unique().tolist()))}
```

```python
import numpy as np

rating_matrix=np.zeros((len(reviews['author_username'].unique().tolist()),len(reviews['movie_id'].unique().tolist())))
rating_matrix.shape
```

```python
for name in reviews['author_username'].unique().tolist():
  tmp=reviews[reviews['author_username']==name]['movie_id'].tolist()
  ratings=reviews[reviews['author_username']==name]['author_rating'].tolist()
  ind=0
  for i in tmp:
    rating_matrix[author_username_voc[name],movie_voc[i]]=ratings[ind]
    ind+=1
```

```python
matrix_df=pd.DataFrame(rating_matrix, columns=reviews['movie_id'].unique().tolist(),index=reviews['author_username'].unique().tolist())
```

```python
matrix_df
```

```python
from sklearn.metrics.pairwise import pairwise_distances

# считаем косинусное расстояние для пользователей и фильмов
user_similarity = pairwise_distances(matrix_df, metric='cosine')
item_similarity = pairwise_distances(matrix_df.T, metric='cosine')
```

```python
import scipy.sparse as sp
from scipy.sparse.linalg import svds
```

```python
r=sp.csr_matrix(matrix_df.to_numpy())
u, s, vt = svds(r, k=10)
vt.shape
X_pred = u @ np.diag(s) @ vt

```

```python
vt.shape
```

```python
def recomend_for_new_user(ratings:dict):
  ids=list(ratings.keys())
  rate=list(ratings.values())
  items=np.where(np.isin(matrix_df.columns, ids))[0]
  item_vecs = vt[:, items].T
  p_new, _, _, _ = np.linalg.lstsq(item_vecs, rate, rcond=None)
  predictions = p_new @ (np.diag(s) @ vt)
  #predictions = np.clip(predictions, 1, 10)
  predictions[items] = -1
  return np.argsort(predictions)[-10:][::-1]
```

```python
rec=recomend_for_new_user({12:10,115:10})
ids=matrix_df.T.iloc[rec].index.tolist()
movies_df[movies_df['id'].isin(ids)]
```

```python
def calculate_ranking_metrics(R_true, R_pred, k=10):
    """
    Метрики ранжирования для топ-k рекомендаций
    """
    n_users = R_true.shape[0]
    precisions = []
    recalls = []
    ndcg_scores = []

    for user in range(n_users):
        true_ratings = R_true[user]
        pred_ratings = R_pred[user]
        relevant_items = np.where(true_ratings > 7)[0]
        if len(relevant_items) == 0:
            continue
        top_k_indices = np.argsort(pred_ratings)[-k:][::-1]
        relevant_in_topk = len(set(top_k_indices) & set(relevant_items))

        precision = relevant_in_topk / k
        precisions.append(precision)


        recall = relevant_in_topk / len(relevant_items)
        recalls.append(recall)


    ranking_metrics = {
        f'Precision@{k}': np.mean(precisions),
        f'Recall@{k}': np.mean(recalls),
    }

    return ranking_metrics
```
